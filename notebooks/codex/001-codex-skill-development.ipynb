## 打印真实目录树

正式开发时，不要先背目录图，先看仓库里真实落了什么。


```python
def print_tree(root: Path, prefix: str = '') -> None:
    entries = sorted(root.iterdir(), key=lambda p: (p.is_file(), p.name))
    for index, entry in enumerate(entries):
        connector = '└── ' if index == len(entries) - 1 else '├── '
        print(prefix + connector + entry.name)
        if entry.is_dir():
            next_prefix = prefix + ('    ' if index == len(entries) - 1 else '│   ')
            print_tree(entry, next_prefix)


print(skill_root)
print_tree(skill_root)
```

## 先记住一个正式开发原则

一份 Skill 不是“目录越多越专业”。

正式开发时，先从最小可用结构开始：

- `SKILL.md`：必需
- `agents/openai.yaml`：推荐
- `references/`：有详细资料时再加
- `scripts/`：有重复动作时再加
- `assets/`：只有真的需要模板或素材时再加

不要为了看起来完整，先创建一堆空目录。


## 先读 `SKILL.md`

`SKILL.md` 是这份 Skill 的核心说明书。

正式开发时，先把这里写准，比先写脚本更重要。


```python
print((skill_root / 'SKILL.md').read_text(encoding='utf-8'))
```

## 看 `SKILL.md` 时应该抓什么

读一份真实 Skill 时，重点抓 4 件事：

1. `name`
2. `description`
3. workflow
4. references / scripts 的使用约定

也就是说，`SKILL.md` 负责回答：

- 这份 Skill 什么时候用
- 这类任务应该按什么步骤做
- 需要时再去读哪些附加资料
- 需要时再跑哪些脚本


```python
from pprint import pprint


skill_reading_focus = {
    'frontmatter': ['name', 'description'],
    'body_focus': ['trigger condition', 'workflow', 'references', 'scripts'],
}

pprint(skill_reading_focus)
```

## 再看 `agents/openai.yaml`

这不是 Skill 的核心逻辑文件，而是推荐的 UI metadata。

正式开发里，它通常负责：

- 展示名称
- 简短描述
- 默认提示词

所以你不应该把工作流细节堆到这里。


```python
print((skill_root / 'agents' / 'openai.yaml').read_text(encoding='utf-8'))
```

## 为什么详细规范要放到 `references/`

一个常见错误是：

- 把所有背景、样例、规则都塞进 `SKILL.md`

这样会让 Skill 越写越臃肿。

更好的正式开发习惯是：

- `SKILL.md` 保持高信号
- 细节下沉到 `references/`
- 只有需要时再读 `references/`


```python
print((skill_root / 'references' / 'notebook_style.md').read_text(encoding='utf-8'))
```

## 为什么重复动作要放到 `scripts/`

另一个常见错误是：

- 每次都手写一遍校验逻辑
- 或者把脚本细节也塞进 `SKILL.md`

正式开发时，更好的方式是：

- 在 `SKILL.md` 里写“何时运行脚本”
- 在 `scripts/` 里放可重复执行的小工具


```python
print((skill_root / 'scripts' / 'validate_notebook.py').read_text(encoding='utf-8'))
```

## 跑一遍真实脚本

教学不能只停留在“看目录”。

正式开发里，你应该验证这份 Skill 自带的脚本真能跑。


```python
import subprocess

result = subprocess.run(
    [
        'python',
        '.agents/skills/openai-notebook-author/scripts/validate_notebook.py',
        'notebooks/openai/002-openai-chat-history.ipynb',
    ],
    capture_output=True,
    text=True,
    check=False,
)

print(result.stdout)
print('returncode =', result.returncode)
```

## 正式开发时，一份 Skill 应该怎么从 0 到 1 落地

这是最重要的一节。

建议按下面顺序做，而不是一上来就造一整套复杂目录。


```python
skill_dev_workflow = [
    'Step 1: 先确认这个任务是不是高频、可复用、值得 Skill 化',
    'Step 2: 先只写 SKILL.md，把触发条件和 workflow 写清楚',
    'Step 3: 如果细节太多，再拆到 references/',
    'Step 4: 如果有重复校验或固定动作，再补 scripts/',
    'Step 5: 如果需要 UI 展示，再补 agents/openai.yaml',
    'Step 6: 用真实任务跑一遍，验证 Skill 说明是否真的够用',
]

for step in skill_dev_workflow:
    print(step)
```

## 结合这个仓库，看这份 Skill 为什么成立

`openai-notebook-author` 值得做成 Skill，不是因为它名字好听，而是因为这个任务真的有稳定 workflow：

1. 先看上一份 notebook
2. 再看学习计划或相关 guide
3. 保持相同教学节奏
4. 新增后做 JSON 校验
5. 更新 README 或学习计划索引

这就是典型的“高频 + 有步骤 + 有仓库约定”的任务。


```python
why_this_is_a_skill = {
    'high_frequency': True,
    'repo_conventions': True,
    'reusable_workflow': True,
    'needs_validation': True,
    'needs_references': True,
}

pprint(why_this_is_a_skill)
```

## 什么情况下不要急着做成 Skill

不是所有东西都值得 Skill 化。

下面这些情况要谨慎：

- 任务只做一次
- 没有稳定流程
- 只是一个简单函数调用
- 只是某个临时命令

换句话说，Skill 不是“把任何需求包装一下”的容器。


```python
not_good_skill_candidates = [
    '一次性临时改一个字段',
    '只需要执行一条简单命令',
    '没有稳定流程的探索性工作',
    '只是调用一次 get_stock_quote(symbol)',
]

for item in not_good_skill_candidates:
    print('-', item)
```

## 如果结合 OpenClaw / Codex 风格来规划教学，建议分 4 课

这次我不建议把教学一开始就铺得很大。

更稳的方式是分层推进。


```python
teaching_plan = [
    {
        'lesson': '第 1 课：读真实 Skill',
        'goal': '看懂一份 Skill 的最小目录和文件职责',
        'artifact': '.agents/skills/openai-notebook-author/',
    },
    {
        'lesson': '第 2 课：写最小 SKILL.md',
        'goal': '只用 frontmatter + workflow 写出最小可用 Skill',
        'artifact': 'SKILL.md',
    },
    {
        'lesson': '第 3 课：补 references 和 scripts',
        'goal': '学会把细节和重复动作拆出去',
        'artifact': 'references/ + scripts/',
    },
    {
        'lesson': '第 4 课：用真实任务验收',
        'goal': '用新增 notebook 这类真实任务验证 Skill 是否真的可用',
        'artifact': 'real task walkthrough',
    },
]

pprint(teaching_plan)
```

## 如果你要自己正式开发一份 Skill，可以直接照这个清单走

这是一个更贴近工程实践的检查表。


```python
skill_checklist = [
    '这个任务是不是会重复出现？',
    '这个任务是不是有稳定步骤？',
    '这个任务是不是依赖仓库或团队约定？',
    'SKILL.md 是否已经写清楚触发条件和 workflow？',
    '详细资料是否已经从 SKILL.md 下沉到 references/？',
    '重复动作是否已经沉淀成 scripts/？',
    '是否用真实任务验证过这份 Skill？',
]

for item in skill_checklist:
    print('[ ]', item)
```

## 当前阶段结论

你现在需要记住：

1. 学 Skill，先看真实目录，再谈抽象概念
2. 正式开发时，先从最小可用结构开始，不要预建一堆空目录
3. `SKILL.md` 负责 workflow，`references/` 放细节，`scripts/` 放重复动作，`agents/openai.yaml` 放 UI metadata
4. Skill 是否成立，关键不在名字，而在这个任务是否高频、可复用、依赖约定
5. 最好的教学不是只讲“Skill 是什么”，而是拿一份真实 Skill 带着跑一遍
